```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

# 1. Carregamento dos Dados
df = pd.read_csv('../data/raw/dataset.csv')
print("--- Primeiras Linhas do Dataset ---")
df.head()
```

```python
# 2. Inspeção Estrutural dos Metadados
print("--- Informações da Base ---")
df.info()

print("\n--- Estatísticas Descritivas ---")
df.describe()
```

```python
# 3. Pré-processamento e Limpeza
# Remoção do identificador único (sem poder preditivo)
df_proc = df.drop(columns=['id_cliente'])

# Codificação da Variável Alvo
le = LabelEncoder()
df_proc['status_emprestimo'] = le.fit_transform(df_proc['status_emprestimo'])

# Separação entre Recursos (X) e Alvo (y)
X = df_proc.drop(columns=['status_emprestimo'])
y = df_proc['status_emprestimo']

# Divisão Estratificada em Treino (80%) e Teste (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Amostras de Treino: {X_train.shape[0]}")
print(f"Amostras de Teste: {X_test.shape[0]}")
```
